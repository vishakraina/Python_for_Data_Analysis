```python
# 1️⃣ Creating Classes and Objects
# Defining a simple class using the `class` keyword
class Rectangle:
    def __init__(self):
        # Attributes (instance variables)
        self.length = 10
        self.breadth = 5

# Creating an object (instance) of the class
rect = Rectangle()

# Accessing attributes using the dot operator
print("Length =", rect.length, "\nBreadth =", rect.breadth)
```

```python
# 🧩 Parameterized Constructor
# Allows assigning values dynamically during object creation
class Rectangle:
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

# Passing parameters to the constructor
rect = Rectangle(40, 20)
print("Length =", rect.length, "\nBreadth =", rect.breadth)
```

```python
# 2️⃣ Class Variables vs Instance Variables
class Circle:
    pi = 3.14  # Class variable shared across all instances

    def __init__(self, radius):
        self.radius = radius  # Instance variable unique per object

circle_1 = Circle(5)
circle_2 = Circle(2)

print("Circle 1: Radius =", circle_1.radius, "\tpi =", circle_1.pi)
print("Circle 2: Radius =", circle_2.radius, "\tpi =", circle_2.pi)
```

```python
# 🔄 Changing a Class Variable

# Changing class variable using the class name affects all objects
Circle.pi = 3.1436

circle_1 = Circle(5)
circle_2 = Circle(2)

print("Circle 1: Radius =", circle_1.radius, "\tpi =", circle_1.pi)
print("Circle 2: Radius =", circle_2.radius, "\tpi =", circle_2.pi)
```

```python
# 3️⃣ Adding Methods to a Class

# Adding a method to calculate area
class Rectangle:
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def calculate_area(self):
        # 'self' gives access to instance attributes
        return self.length * self.breadth

rect = Rectangle(10, 5)
print("Length =", rect.length, "\nBreadth =", rect.breadth, "\nArea =", rect.calculate_area())
```

```python
# ✳️ Significance of self

# Each object maintains its own data using 'self'
new_rect = Rectangle(15, 8)
print("New Rectangle: Length =", new_rect.length, "\nBreadth =", new_rect.breadth, "\nArea =", new_rect.calculate_area())

print("Original Rectangle: Length =", rect.length, "\nBreadth =", rect.breadth, "\nArea =", rect.calculate_area())
```

```python
# 4️⃣ Inheritance and Method Overriding
# Basic Inheritance Example
class Employee:
    def function1(self):
        print("Hello World")

# Department inherits Employee
class Department(Employee):
    pass

emp = Employee()
dept = Department()

emp.function1()
dept.function1()

# ⚙️ Method Overriding Example
class Shape:
    def set_color(self, color):
        self.color = color

    def calculate_area(self):
        pass  # Placeholder to be overridden

    def color_the_shape(self):
        color_price = {"red": 10, "blue": 15, "green": 5}
        return self.calculate_area() * color_price[self.color]


class Circle(Shape):
    pi = 3.14

    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        return Circle.pi * self.radius ** 2


class Rectangle(Shape):
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def calculate_area(self):
        return self.length * self.breadth

    def __str__(self):
        return "Area of rectangle = " + str(self.calculate_area())


# Object creation
c = Circle(5)
c.set_color("red")
print("Circle with radius =", c.radius, "when colored", c.color, "costs $", c.color_the_shape())

r = Rectangle(5, 10)
r.set_color("blue")
print("Rectangle with length =", r.length, "and breadth =", r.breadth, "when colored", r.color, "costs $", r.color_the_shape())
```

```python
# 5️⃣ Polymorphism
# Different classes share a common interface (method names)
class Bird:
    def intro(self):
        print("There are many types of birds.")

    def flight(self):
        print("Most of the birds can fly but some cannot.")


class Sparrow(Bird):
    def flight(self):
        print("Sparrows can fly.")


class Ostrich(Bird):
    def flight(self):
        print("Ostriches cannot fly.")


# Demonstrating Polymorphism
obj_bird = Bird()
obj_spr = Sparrow()
obj_ost = Ostrich()

obj_bird.intro()
obj_bird.flight()
obj_spr.intro()
obj_spr.flight()
obj_ost.intro()
obj_ost.flight()
```

```python
# 6️⃣ Encapsulation
# Encapsulation: Protecting data using protected and private variables
class EncapsulatedClass:
    def __init__(self):
        self._protected_var = 10   # Protected variable (accessible with caution)
        self.__private_var = 20    # Private variable (name mangling applies)

    def get_private_var(self):
        # Getter method to access private variable
        return self.__private_var

    def set_private_var(self, value):
        # Setter method to modify private variable safely
        self.__private_var = value


obj = EncapsulatedClass()

# Accessing protected and private variables
print("Protected variable:", obj._protected_var)
print("Private variable:", obj.get_private_var())

# Updating private variable using setter
obj.set_private_var(30)
print("Updated private variable:", obj.get_private_var())
```
