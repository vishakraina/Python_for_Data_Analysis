```python
# Matplotlib and Pandas Visualization Guide

## 1. Creating a DataFrame for Visualization

#Matplotlib works well with Pandas DataFrames, which are often used to prepare data for visualization.

#> **Example: Creating a DataFrame from a Dictionary**

# Import Pandas
import pandas as pd
# Create dictionary data
Data = {
    'Year': [1920, 1930, 1940, 1950, 1960, 1970, 1980, 1990, 2000, 2010, 2020],
    'Exchange Rate': [65, 69, 71, 64, 62, 59, 72, 71, 75, 78, 81]
}

# Convert dictionary to DataFrame
df = pd.DataFrame(Data)

# Display first 5 rows and data info
print(df.head(5))
print("Type of Data:", type(Data))
print("Type of df:", type(df))
print(df.info())

#> **Notes:**
#> - The dictionary is converted to a Pandas DataFrame using `pd.DataFrame()`.
#> - Explicitly specifying column names is optional if dictionary keys match column names.
#> - Mis-specified column names may create empty columns with NaN values.
```

```python
import matplotlib.pyplot as plt

## 2. Plotting with Matplotlib and Pandas

#Pandas provides a convenient `.plot()` method that integrates with Matplotlib.

### Line Plot

df.plot(x='Year', y='Exchange Rate', kind='line')
plt.show()
# Note: Shows exchange rate trend over years.
```

```python
### Area Plot

df.plot(x='Year', y='Exchange Rate', kind='area')
plt.show()
# Note: Fills area under the exchange rate curve.
```

```python
### Bar Plot

df.plot(x='Year', y='Exchange Rate', kind='bar')
plt.show()
# Note: Compares exchange rate per year using vertical bars.
```

```python
### Horizontal Bar Plot

df.plot(x='Year', y='Exchange Rate', kind='barh')
plt.show()
# Note: Similar to bar plot but oriented horizontally.
```

```python
### Scatter Plot

df.plot(x='Year', y='Exchange Rate', kind='scatter')
plt.show()
# Note: Displays data points showing relation between year and rate.
```

```python
### Customizing Scatter Plot

import numpy as np
N = 11
colors = np.random.rand(N)
plt.scatter(df['Year'], df['Exchange Rate'], c=colors)
plt.show()
# Note: Randomly colors each data point.
```

```python
## 3. Pie Charts

#Pie charts show category proportions.

### Simple Pie Chart

Data = {'Tasks': [100, 500, 300]}
df2 = pd.DataFrame(Data, columns=['Tasks'], index=['Pending', 'Completed', 'Ongoing'])
df2.plot.pie(y='Tasks', figsize=(5, 5))
plt.show()
# Note: Shows distribution of task statuses.
```

```python
### Customized Pie Chart

labels = ['Java', 'Python', 'R', 'Javascript']
sizes = [15, 30, 45, 10]
explode_labels = (0, 0.2, 0, 0)
fig1, ax1 = plt.subplots()
ax1.pie(sizes, explode=explode_labels, labels=labels, shadow=True, startangle=90)
ax1.axis('equal')
plt.show()
# Note: Exploded Python slice and adds shadow for visual appeal.
```

```python
## 4. Working with a Real Dataset (Churn Modelling)

### Loading and Cleaning Data

churn_df = pd.read_csv('C:\\Users\\rainavi\\Downloads\\Coding\\Pyhton for Data Analytics\\Churn_Modelling.csv')
print(churn_df.head(5))
print(churn_df.tail(5))
# Note: Load and inspect dataset.
```

```python
### Task 1: Remove Unnecessary Columns

churn_df.drop(['RowNumber', 'CustomerId', 'Surname'], axis=1, inplace=True)
print(churn_df.head(5))
# Note: Simplifies dataset by removing ID columns.
```

```python
### Task 2: Count Values in Categorical Column

print(churn_df['Geography'].value_counts())
# Note: Shows frequency count per geography.
```

```python
### Task 3: Calculate Percentage of Each Category

print(churn_df['Geography'].value_counts() / len(churn_df) * 100)
# Note: Displays category distribution in percentages.
```

```python
### Task 4: Plot Geography Counts

churn_df['Geography'].value_counts().plot(kind='bar')
plt.show()

churn_df['Geography'].value_counts().plot(kind='barh')
plt.show()

churn_df['Geography'].value_counts().plot(kind='pie')
plt.show()
# Note: Compare data distribution in multiple formats.
```

```python

### Task 5: Distribution of Exited Column

print(churn_df['Exited'].value_counts() / len(churn_df) * 100)
churn_df['Exited'].value_counts().plot(kind='bar')
plt.show()
# Note: Visualize customer retention vs churn percentage.
```

```python
## 5. Advanced Visualizations

### Scatter Plot (Age vs Tenure)

plt.scatter(churn_df['Age'], churn_df['Tenure'])
plt.title('Scatter Plot')
plt.xlabel('Age')
plt.ylabel('Tenure')
plt.show()
# Note: Shows relationship between customer age and tenure.
```

```python
### Histogram (Tenure Distribution)

plt.hist(churn_df['Tenure'], bins=30)
plt.title('Histogram')
plt.xlabel('Tenure')
plt.ylabel('Frequency')
plt.show()
# Note: Displays distribution of tenure data.
```

```python
### Bar Plot (Mean Estimated Salary by Geography)

plt.bar(churn_df['Geography'], churn_df['EstimatedSalary'].mean(), color='cyan', edgecolor='orange')
plt.title('Geography vs Estimated Salary', fontdict={'fontsize': 20, 'color': 'Green'})
plt.show()
# Note: Compares mean estimated salary per region.
```

```python
### Box Plot (Age Distribution)

churn_df['Age'].plot.box()
plt.show()
print(churn_df['Age'].describe())
# Note: Box plot shows quartiles, median, and outliers in age distribution.
```

```python
## 6. Subplots

labels_ex = ['JavaScript', 'Java', 'Python', 'R']
sizes = [15, 30, 45, 10]
explode_labels = (0, 0.2, 0, 0)
fig1, ax1 = plt.subplots()
ax1.pie(sizes, explode=explode_labels, labels=labels_ex, shadow=True, startangle=90)
ax1.axis('equal')
plt.show()
# Note: Displays exploded pie chart as subplot.
```
