# Pandas Library in Python

## Introduction to Pandas
Pandas is a fast, powerful, flexible, and easy-to-use open-source Python library designed for data analysis and manipulation. 
It is widely used in data science and analytics for handling structured data, such as tabular data in the form of DataFrames and Series.

### Why Use Pandas?
- **Speed**: Efficiently processes large datasets with optimized performance.
- **Flexibility**: Supports a wide8 range of data formats (CSV, Excel, JSON, etc.) and operations.
- **Ease of Use**: Provides intuitive methods for data manipulation and analysis.
- **Data Science**: Essential for data cleaning, exploration, and statistical analysis.

````python
## 1. Creating and Reading DataFrames
# A DataFrame is a tabular data structure in Pandas, similar to a spreadsheet or SQL table, with rows and columns. 
# Pandas can read data from various file formats like CSV, Excel, and JSON.

# Read a CSV file
df = pd.read_csv("C:\\Users\\rainavi\\Downloads\\Coding\\Pyhton for Data Analytics\\Churn_Modelling.csv")

# Display first few rows
print(df.head())

# Display metadata
print(df.info())

# Statistical summary
print(df.describe())
````

````python
## 2. Accessing and Modifying Data

# Accessing columns
print(df['Balance'])              # Access a single column
print(df[['CustomerId', 'Age']])  # Access multiple columns

# Accessing rows
print(df.iloc[0:5])               # Access rows by position
print(df.loc[0:5])                # Access rows by label

# Creating new columns
df['NewSalary'] = df['EstimatedSalary'] * 1.1
df['FullName'] = df['CustomerId'].astype(str) + ' ' + df['Surname']
df['Bal_SQRT'] = df['Balance'].apply(np.sqrt)
````

````python
## 3. Filtering and Sorting

# Filtering data
filtered_df = df[df['Age'] >= 50]

# Sorting data
df_sorted = df.sort_values(by=['Age', 'Tenure'], ascending=[False, True])
````

````python
## 4. Handling Missing Values
dfa = pd.read_csv("C:\\Users\\rainavi\\Downloads\\Coding\\Pyhton for Data Analytics\\Test.csv")
# Check missing values
print(dfa.info())

# Drop rows with missing values
dfa_clean = dfa.dropna()
print(df_clean)

# Fill missing values
dfa['Age'] = dfa['Age'].fillna(dfa['Age'].median())
print(dfa)
````

````python
## 5. Removing Columns
df = pd.read_csv("C:\\Users\\rainavi\\Downloads\\Coding\\Pyhton for Data Analytics\\Churn_Modelling.csv")

# Remove a column using pop()
df.pop('RowNumber')

# Remove columns using drop()
df = df.drop(columns=['Surname', 'CreditScore'])

print(df)

````

````python
# Group by Geography and compute mean
geo_mean = df.groupby('Geography').mean(numeric_only =True)
print(geo_mean)

# Group by multiple columns
geo_gender_mean = df.groupby(['Geography', 'Gender'])['Balance'].mean()
print(geo_gender_mean)
````

````python
## 7. Combining DataFrames

# Concatenation
df1 = pd.DataFrame({'A': ['A0', 'A1'], 'B': ['B0', 'B1']})
print(df1)
df2 = pd.DataFrame({'A': ['A2', 'A3'], 'B': ['B2', 'B3']})
print(df2)
result = pd.concat([df1, df2])
print(result)
# Merging (Joins)
df1 = pd.DataFrame({'cust_id': [1, 2, 3], 'name': ['A', 'B', 'C']})
print(df1)
df2 = pd.DataFrame({'cust_id': [2, 3, 4], 'city': ['X', 'Y', 'Z']})
print(df2)
merged = pd.merge(df1, df2, on='cust_id', how='inner')
print(merged)
````
